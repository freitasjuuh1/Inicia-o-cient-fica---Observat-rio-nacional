# Princípios fundamentais do método Magnetotelúrico (MT)
As equações fundamentais que descrevem o comportamento dos campos eletromagnéticos são as quatro equações de Maxwell, elas podem ser expressas na forma diferencial como:

* Lei de Gauss para o campo elétrico

   * **∇⋅𝐸=𝜌/𝜀0​**


* Lei de Gauss para o campo magnético

   * **∇⋅𝐵=0**
  

* Lei de Faraday da indução

    * **∇×E=−∂B/∂t​**

* Lei de Ampère-Maxwell

    * **∇×𝐵= μ0𝐽 + 𝜇0𝜀0*∂𝐸/∂𝑡**



A partir dessas equações, e considerando os pressupostos físicos do método magnetotelúrico (MT), é possível obter as equações de difusão eletromagnética, as quais permitem a interpretação das respostas da Terra a fontes eletromagnéticas naturais, relacionando as variações temporais dos campos elétrico e magnético às propriedades elétricas do subsolo, em particular à condutividade elétrica.

Considerando que as variações temporais dos campos eletromagnéticos naturais podem ser descritas como superposições de componentes harmônicas, adota-se uma dependência temporal do tipo
𝑒^𝑖𝜔𝑡. Essa abordagem permite formular o problema no domínio da frequência, aplica-se então a transformada de fourrier, no qual ∂/∂t​=iω.  Dessa forma, as equações de Maxwell assumem uma forma mais simples, tornando possível a obtenção da equação de difusão eletromagnética que governa o comportamento dos campos no método magnetotelúrico.

# As considerações do método MT

Considerando as leis fundamentais que regem o método magnetotelúrico (MT), torna-se necessário adotar algumas aproximações físicas que permitem simplificar as equações de Maxwell e torná-las tratáveis do ponto de vista analítico e numérico. Essas aproximações são justificadas pelas características do regime de frequências envolvido e pelas propriedades eletromagnéticas do meio terrestre


1.   Assumindo que não existam fontes de corrente dentro da Terra, podemos considerar:

                                             ∇⋅𝐸=0

2.   Deve se considerar também que no regime quase-estático adotado no método magnetotelúrico, as frequências envolvidas são suficientemente baixas para que a corrente de deslocamento seja muito menor que a corrente de condução no meio terrestre, sendo assim:

                                        ∇×𝐵=𝜇0* 𝜀0* ∂𝐸/∂𝑡

Além disso as equações de Maxwell são complementadas pelas seguintes equações constitutivas:

                                      D=𝜀0*E

                                      B=𝜇0*H

                                      J= σ*E

A partir dessas considerações temos que as quatro equações no dominío da Terra serão:

*Lei de Gauss para o campo elétrico*

∇⋅𝐸=0

*Lei de Gauss para o campo magnético*

∇⋅𝐵=0

*Lei de Faraday da indução*

∇×E=−∂B/∂t

*Lei de Ampère-Maxwell*

∇×𝐵=𝜇0* σ* E

# A equação de difusão

A partir da lei de Faraday da indução

                          ∇×E=−∂B/∂t

 aplica-se o rotacional dos dois lados da equação, obtendo:

                       ∇x(∇×E)= ∇×(- ∂B/∂t)

Nota-se o surgimento da identidade vetorial no lado esquerdo da equação

                                           ∇×(∇×E)

a qual pode ser expressa como

                                    ∇×(∇×E)=∇(∇⋅E)−∇^2 E.

Temos então:

                                ∇(∇⋅E)−∇^2 E= -∂(∇×B)/∂t

Sabendo que:

                                      ∇⋅𝐸=0 e ∇×𝐵=𝜇0* σ* E

Logo:

                                ∇(0)−∇^2 E= -∂(𝜇0* σ* E)/∂t*

A equação de difusão do campo elétrico pode ser expressa como:

                                ∇^2 E= 𝜇0* σ* ∂E/∂t

Está equação está no dominío temporal, para coloca-lá no dominío da frequência basta utilizar a transformada de Fourrier explicita acima

                                ∂/∂t​=iω
Ou seja:  ∂E/∂t​=iωE, logo:

                              ∇^2 E= 𝜇0* σ* i *ω *E



#A solução da equação

trata-se de uma equação diferencial parcial de segunda ordem com coeficientes constantes. A solução geral é do tipo exponencial complexa, podendo ser expressa como

                              E(z)=Ae^−kz +Be^kz

onde k é o número de onda difusivo descrito por:

                                k=(iωμ0​σ)^1/2​

No contexto físico do método magnetotelúrico, considera-se apenas a solução que decai com a profundidade, uma vez que os campos eletromagnéticos se atenuam à medida que penetram no meio condutor. Assim, o termo exponencial crescente é descartado, resultando em:


                                E(z)=E0	​e^(−kz)






```python
import numpy as np
import matplotlib.pyplot as plt

#Fequência
f1 = 10 # Hz
f2 = 1 # Hz
f3 = 0.1

omega1 =2*np.pi*f1
omega2 =2*np.pi*f2
omega3 = 2*np.pi*f3

#Dominíos da porfundidade:
z = np.linspace(0, 100000, 1000)

#Constantes
E0=1.0
mu0=4*np.pi*1e-7
sigma1= 1*1e-2
sigma2= 1*1e-1
sigma3= 1*1e0
```

```python
# número de onda difusivo
k1 = np.sqrt(1j * omega1 * mu0 * sigma1) #
k2 = np.sqrt(1j * omega2 * mu0 * sigma1) # mantém a mesma condutividade e muda a frequência

#Campo elétrico
E1 = E0 * np.exp(-k1*z)
E2 = E0 * np.exp(-k2*z)
```

```python
# @title Texto de título padrão
figsize=(10, 5)
plt.plot(np.imag(E1), z, label=f"Im(E(z)) (f={f1} Hz, $\sigma$={sigma1} S/m)", color='red')
plt.ylabel("Profundidade z (m)")
plt.xlabel("Im(E(z)) (V/m)")
plt.title(f"Campo Elétrico")
plt.ylim(0, 50000)
plt.grid(True)
plt.gca().invert_yaxis() # Invert y-axis for depth
plt.tick_params(axis='x', colors='red')
plt.gca().xaxis.label.set_color('red')

plt.twiny()
plt.plot(np.real(E1), z, label=f"Re(E(z)) (f={f1} Hz, $\sigma$={sigma1} S/m)", color='blue', linestyle='--')
plt.xlabel("Re(E(z)) (V/m)")
plt.tick_params(axis='x', colors='blue')
plt.gca().xaxis.label.set_color('blue')
```

#Skin depth
O skin depth (profundidade de penetração) é a escala característica que controla a profundidade efetiva de investigação do método magnetotelúrico, dependendo essencialmente da frequência (ou período) do sinal eletromagnético e da condutividade elétrica do meio. Dessa forma, o método possibilita a investigação de diferentes profundidades do subsolo. Esse princípio descreve o decaimento exponencial dos campos eletromagnéticos à medida que se difundem em um meio condutor.

#A equação

Considerando a solução da equação de difusão do campo elétrico:

    E(z)=E0 ​e^(−kz)

em que o número de onda é definido por:

    k=(i)^1/2*(ωμ0​σ)^1/2​

e levando em conta a definição de raiz quadrada no plano complexo:

    (i)^1/2= 1 + i/(2)^1/2

Temos então:

    (i)^1/2= k/(ωμ0​σ)^1/2​

Em seguida:

    (k/(ωμ0​σ)^1/2​)= 1+i/(2)^1/2

Logo:

    k=((1+i) * (ωμ0​σ)^1/2)/ (2)^1/2

Sabendo que o skin depht é inversamente proporcional a parte real do número  de onda difusivo, podemos concluir que:

    δ=(2/ωμ0​σ)^1/2





```python
def plot_e_field(f_val, sigma_val, z_val, ax, title_suffix=''):
    omega_val = 2 * np.pi * f_val
    k_val = np.sqrt(1j * omega_val * mu0 * sigma_val)
    E_val = E0 * np.exp(-k_val * z_val)

    # Plotting Imaginary part on the primary (bottom) x-axis
    ax.plot(np.imag(E_val), z_val, label=f"Im(E(z)) (f={f_val} Hz, $\sigma$={sigma_val} S/m)", color='red')
    ax.set_ylabel("Profundidade z (m)")
    ax.set_xlabel("Im(E(z)) (V/m)")
    ax.set_title(f"Campo Elétrico {title_suffix}")
    ax.set_ylim(0, 50000)
    ax.grid(True)
    ax.invert_yaxis() # Invert y-axis for depth
    ax.tick_params(axis='x', colors='red')
    ax.xaxis.label.set_color('red')

    # Create a second x-axis for the Real part
    ax2 = ax.twiny()
    ax2.plot(np.real(E_val), z_val, label=f"Re(E(z)) (f={f_val} Hz, $\sigma$={sigma_val} S/m)", color='blue', linestyle='--')
    ax2.set_xlabel("Re(E(z)) (V/m)")
    ax2.tick_params(axis='x', colors='blue')
    ax2.xaxis.label.set_color('blue')

    lines, labels = ax.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax.legend(lines + lines2, labels + labels2, loc='lower right', fontsize='small')

    # Combine legends from both axes
    lines, labels = ax.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax.legend(lines + lines2, labels + labels2, loc='lower right', fontsize='small')

    ## Add annotation for frequency and conductivity
    ax.annotate(f'f = {f_val} Hz\n$\sigma$ = {sigma_val} S/m', xy=(0.45, 0.25), xycoords='axes fraction', fontsize=10, bbox=dict(boxstyle="round,pad=0.3", fc="yellow", ec="b", lw=1, alpha=0.5))

# Create a figure with a 2x3 grid of subplots
fig, axes = plt.subplots(2, 3, figsize=(15, 10)) # Adjust figure size for 2x3 grid

# Flatten the axes array for easier iteration
axes = axes.flatten()

# Row 1: Varying frequency, constant sigma1
plot_e_field(f1, sigma1, z, axes[0], title_suffix='(f=10Hz, \u03c3=0.01)')
plot_e_field(f2, sigma1, z, axes[1], title_suffix='(f=1Hz, \u03c3=0.01)')
plot_e_field(f3, sigma1, z, axes[2], title_suffix='(f=0.1Hz, \u03c3=0.01)')

# Row 2: Constant frequency (f2), varying sigma
plot_e_field(f2, sigma3, z, axes[3], title_suffix='(f=1Hz, \u03c3=1)')
plot_e_field(f2, sigma2, z, axes[4], title_suffix='(f=1Hz, \u03c3=0.1)')
plot_e_field(f2, sigma1, z, axes[5], title_suffix='(f=1Hz, \u03c3=0.01)')

plt.tight_layout()
plt.show()
```

#Tipper no método MT

No método Magnetotelúrico (MT), existe uma relação fundamental que conecta as componentes horizontais do campo magnético à sua componente vertical. Essa relação é descrita pelos vetores de indução, mais conhecidos como tipper. O tipper é uma quantidade complexa, contendo partes real e imaginária essa relação pode ser representada a partir da seguinte equação:

    Hz(ω)= Tx(ω).Hx(ω) + Ty(ω).Hy(ω)

Existem duas convenções clássicas para a representação dos vetores de indução. Na convenção de Parkinson (1959), eles apontam diretamente para as regiões de concentração interna de corrente, indicando a presença de um condutor elétrico no subsolo. Já na convenção de Wiese (1962), os vetores são definidos com sentido oposto, apontando para longe da concentração de corrente. Embora matematicamente equivalentes, essas duas convenções diferem apenas no sinal adotado na definição do tipper.

#Em meios geológicos

Quando se leva em consideração os tipos de meios existentes (1D, 2D e 3D), o comportamento das setas de indução apresenta diferenças fundamentais. Em um meio unidimensional (1D), no qual a condutividade elétrica varia apenas com a profundidade, e considerando que o tipper é sensível à presença de condutores laterais, a componente vertical do campo magnético é nula, resultando em vetores de indução iguais a zero.

Já em meios bidimensionais (2D) e tridimensionais (3D), nos quais existe variação lateral de condutividade, a função de transferência geomagnética assume valores diferentes de zero, apresentando magnitudes relacionadas às concentrações de corrente elétrica no subsolo. Apesar disso, os vetores de indução exibem comportamentos distintos dependendo da dimensionalidade do meio.

Em um subsolo bidimensional, os vetores de indução não nulos tendem a se alinhar perpendicularmente ao strike geológico, refletindo a geometria lateral da estrutura condutiva. Por outro lado, em um meio tridimensional, as setas de indução não apresentam uma direção preferencial bem definida, evidenciando a complexidade das heterogeneidades condutivas presentes
